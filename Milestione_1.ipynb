# Milestone 1: Data Collection & Preprocessing

## Dataset Overview

The dataset contains historical visa application records including:

- Case received date
- Decision date
- Country of citizenship
- Class of admission (visa type)
- Employer details
- Wage and prevailing wage details

The target variable for this project is `processing_days`, which represents the number of days taken to process a visa application.

```python
col_needed= [
    "case_received_date",
    "decision_date",
    "case_status",
    "application_type",
    "refile",
    "class_of_admission",
    "country_of_citizenship",
    "foreign_worker_info_birth_country",
    "foreign_worker_info_education",
    "foreign_worker_info_major",
    "foreign_worker_info_rel_occup_exp",
    "foreign_worker_info_req_experience",
    "foreign_worker_yr_rel_edu_completed",
    "foreign_worker_ownership_interest",
    "employer_state",
    "employer_country",
    "employer_num_employees",
    "employer_yr_estab",
    "job_info_job_title",
    "job_info_work_state",
    "job_info_work_city",
    "job_info_education",
    "job_info_major",
    "job_info_experience",
    "job_info_experience_num_months",
    "job_info_foreign_lang_req",
    "pw_level_9089",
    "pw_soc_code",
    "pw_soc_title",
    "pw_amount_9089",
    "pw_unit_of_pay_9089",
    "wage_offer_from_9089",
    "wage_offer_to_9089",
    "wage_offer_unit_of_pay_9089",
    "naics_code",
    "naics_title",
    "us_economic_sector"
]

df = pd.read_csv("D:/visa_predictor/archive_1/us_perm_visas.csv", usecols=col_needed)
```

```python
cols_to_drop = [
    "wage_offer_to_9089",
    "foreign_worker_yr_rel_edu_completed",
    "naics_title",
    "naics_code",
    "foreign_worker_info_birth_country",
    "foreign_worker_info_rel_occup_exp",
    "foreign_worker_info_req_experience",
    "foreign_worker_ownership_interest",
    "refile",
    "us_economic_sector",
    "job_info_experience_num_months",
    "application_type"
]
df.drop(columns=cols_to_drop, inplace=True)
```

```python
missing_percent = (df.isnull().mean() * 100).sort_values(ascending=False)
missing_percent
```

```python
numeric_cols = [
    "job_info_experience",
    "employer_yr_estab",
    "employer_num_employees",
    "wage_offer_from_9089",
    "pw_amount_9089"
]


for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

for col in numeric_cols:
    median_val = df[col].median()
    df[col] = df[col].fillna(median_val)

```

```python
categorical_cols = [
    "job_info_major",
    "job_info_education",
    "foreign_worker_info_major",
    "foreign_worker_info_education",
    "job_info_foreign_lang_req",
    "pw_level_9089",
    "class_of_admission",
    "country_of_citizenship",
    "pw_soc_title",
    "pw_unit_of_pay_9089",
    "pw_soc_code",
    "job_info_work_state",
    "job_info_work_city",
    "employer_state",
    "case_status"
]

for col in categorical_cols:
    mode_val = df[col].mode()[0]
    df[col] = df[col].fillna(mode_val)
```

```python
df["case_received_date"] = pd.to_datetime(df["case_received_date"], errors='coerce')
df["decision_date"] = pd.to_datetime(df["decision_date"], errors='coerce')

df = df.dropna(subset=["case_received_date", "decision_date"])
```

```python
df = df.copy() 
flag_cols = ["job_info_foreign_lang_req", "pw_level_9089"]
for col in flag_cols:
    df.loc[:, col] = df[col].map({"Y": 1, "N": 0}).astype(float)
    df.loc[:, col] = df[col].fillna(0)


```

```python
missing_per_col = df.isnull().sum()
print(missing_per_col)
```

```python
df.info()
```

```python
# handling missing values
df["employer_country"] = df["employer_country"].fillna("Unknown")
df["job_info_job_title"] = df["job_info_job_title"].fillna("Unknown")


df["job_info_experience"] = df["job_info_experience"].map({"Y": 1, "N": 0}).fillna(0)


df["wage_offer_unit_of_pay_9089"] = pd.to_numeric(df["wage_offer_unit_of_pay_9089"], errors='coerce').fillna(0)
```

## Target Variable Creation

Processing time is calculated as the difference (in days) between the visa decision date and the application submission date.

This variable will act as the prediction target for future modeling.

```python
# Calculate processing days
df["processing_days"] = (df["decision_date"] - df["case_received_date"]).dt.days
```

```python
df = df[df['processing_days'] >= 0]

# Check summary statistics
df['processing_days'].describe()
```

All negative or invalid processing times were removed to ensure data consistency.
